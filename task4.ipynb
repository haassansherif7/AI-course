**4-Write a Python function that takes a list of words and return the longest word and the length of the longest one
(Longest word: Exercises

```python
def long():
    A = []
    N = int(input("Enter number of words: "))

    for i in range(N):
        word = input(f"Enter word {i + 1}: ")
        A.append(word)

    longest = max(A, key=len)

    print(f"Longest word: {longest}")
    print(f"Length of the longest word: {len(longest)}")

long()
```

```python
text = input("Enter a string")

count = {}

for char in text:
    if char in count:
        count[char] = count[char] + 1
    else:
        count[char] = 1

print("repeated characters only:")
for char in count:
    if count[char] > 1:
        print(char, "=", count[char])

for char in text:
    if count[char] == 1:
        print("first non-repeating character is:", char)
        break
```

**31-write a python program Extract all elements with Frequency greater than K (Input : test_list = [4, 6, 4, 3, 3, 4, 3, 4, 3, 8], K = 3
Output : [4, 3] )

```python
A = []
N = int(input("Enter number of elements: "))
for i in range(N):
    value = int(input(f"Enter element {i + 1}: "))
    A.append(value)

print("List:", A)

K = int(input("Enter frequency limit: "))

frequency = {}

for i in A:
    if i in frequency:
        frequency[i] += 1
    else:
        frequency[i] = 1
R=[]
for l in frequency:
    if frequency[l] > K:
        R.append(l)
print(f"Elements with frequency greater than {K} are:",R)
```

**39-write a Python Program to Sort the list according to the column using lambda
array = [[1, 3, 3], [2, 1, 2], [3, 2, 1]]
Output :
Sorted array specific to column 0, [[1, 3, 3], [2, 1, 2], [3, 2, 1]]
Sorted array specific to column 1, [[2, 1, 2], [3, 2, 1], [1, 3, 3]]
Sorted array specific to column 2, [[3, 2, 1], [2, 1, 2], [1, 3, 3]]

```python
array = [[1, 3, 3], [2, 1, 2], [3, 2, 1]]

column = 1

n = len(array)

for i in range(n):
    for j in range(0, n - i - 1):
        if array[j][column] > array[j + 1][column]:
            array[j], array[j + 1] = array[j + 1], array[j]

print(f"Sorted by column {column}:", array)
```

**40- write a program to Sort Python Dictionaries by Key or Value
Input:
{'ravi': 10, 'rajnish': 9, 'sanjeev': 15, 'yash': 2, 'suraj': 32}

Output:
{'rajnish': 9, 'ravi': 10, 'sanjeev': 15, 'suraj': 32, 'yash': 2}

```python
A = {'ravi': 10, 'rajnish': 9, 'sanjeev': 15, 'yash': 2, 'suraj': 32}
A_items = list(A.items())

for i in range(len(A_items)):
    for j in range(i + 1, len(A_items)):
        if A_items[i][1] > A_items[j][1]:
            A_items[i], A_items[j] = A_items[j], A_items[i]


sorted_A = dict(A_items)

print("Sorted Dictionary by Value:", sorted_A)
```
